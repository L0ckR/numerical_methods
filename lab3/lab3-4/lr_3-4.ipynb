По трём соседним точкам строим квадратичный интерполяционный многочлен и дифференцируем его в x*=0.2.

## Исходные данные

Числа ниже соответствуют файлу input.txt этой работы.

```python
import math
import numpy as np
import matplotlib.pyplot as plt
from typing import List

rows = [[-0.2, 0.0, 0.2, 0.4, 0.6], [-0.20136, 0.0, 0.20136, 0.41152, 0.6435], [0.2]]
xs, ys, x_star = rows[0], rows[1], rows[2][0]
print('Точка:', x_star)
```

## Первая и вторая производные

Разделённые разности задают коэффициенты локального многочлена Ньютона. Его производные служат приближением производных табличной функции.

```python
def df(x_test, x, y):
    for i in range(len(x) - 2):
        if x[i] <= x_test <= x[i + 1]:
            break
    a1 = (y[i + 1] - y[i]) / (x[i + 1] - x[i])
    a2 = ((y[i + 2] - y[i + 1]) / (x[i + 2] - x[i + 1]) - a1)
    a2 = a2 / (x[i + 2] - x[i]) * (2 * x_test - x[i] - x[i + 1])
    return a1 + a2


def d2f(x_test, x, y):
    for i in range(len(x) - 2):
        if x[i] <= x_test <= x[i + 1]:
            break
    num = (y[i + 2] - y[i + 1]) / (x[i + 2] - x[i + 1]) - (y[i + 1] - y[i]) / (x[i + 1] - x[i])
    return 2 * num / (x[i + 2] - x[i])
```

```python
first = df(x_star, xs, ys)
second = d2f(x_star, xs, ys)
print('f′(x*) =', first)
print('f″(x*) =', second)
```

## Самопроверка

Поскольку узлы равномерны, сравни с центральными разностями.

```python
i = xs.index(x_star)
h = xs[i+1]-xs[i]
central_first = (ys[i+1]-ys[i-1])/(2*h)
central_second = (ys[i+1]-2*ys[i]+ys[i-1])/h**2
assert abs(first-central_first) < 1e-12
assert abs(second-central_second) < 1e-12
print('Центральные разности:', central_first, central_second)
```
